```python
import pandas as pd

# Load dataset
df = pd.read_csv("../dataset/loan_approval_dataset.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Display first 5 rows
df.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,x="loan_status")

plt.title("Loan Status Distribution")

plt.savefig("../outputs/graphs/loan_status.png")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,x="education")

plt.title("Education Distribution")

plt.savefig("../outputs/graphs/education.png")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,x="self_employed")

plt.savefig("../outputs/graphs/self_employed.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["income_annum"],bins=30)

plt.title("Annual Income Distribution")

plt.savefig("../outputs/graphs/income.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["loan_amount"],bins=30)

plt.savefig("../outputs/graphs/loan_amount.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["cibil_score"],bins=25)

plt.savefig("../outputs/graphs/cibil_score.png")

plt.show()
```

```python
plt.figure(figsize=(10,8))

numeric_df = df.select_dtypes(include="number")

sns.heatmap(numeric_df.corr(),annot=True,cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.savefig("../outputs/graphs/heatmap.png")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.boxplot(y=df["loan_amount"])

plt.savefig("../outputs/graphs/boxplot.png")

plt.show()
```

```python
sns.pairplot(df[[
    "income_annum",
    "loan_amount",
    "cibil_score"
]])

plt.savefig("../outputs/graphs/pairplot.png")

plt.show()
```
